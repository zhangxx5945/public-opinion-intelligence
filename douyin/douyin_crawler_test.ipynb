# 抖音单视频信息提取测试

Notebook 负责配置和查看结果，实际抓取逻辑位于 `extract_one_video.py`。

```python
from pathlib import Path
import json
import subprocess
import sys

VIDEO_URL = "https://www.douyin.com/jingxuan?modal_id=7654088004834757922"
SCRIPT_PATH = Path.cwd() / "extract_one_video.py"
OUTPUT_PATH = Path.cwd() / "douyin_video_results.json"

if not SCRIPT_PATH.exists():
    raise FileNotFoundError(f"找不到脚本：{SCRIPT_PATH}")
```

```python
completed = subprocess.run(
    [
        sys.executable,
        str(SCRIPT_PATH),
        VIDEO_URL,
        "--output",
        str(OUTPUT_PATH),
    ],
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
    timeout=90,
)

if completed.returncode != 0:
    raise RuntimeError(completed.stderr.strip() or "抓取程序运行失败")

video = json.loads(completed.stdout)
video
```

```python
videos = json.loads(OUTPUT_PATH.read_text(encoding="utf-8"))
print(f"当前共保存 {len(videos)} 条视频记录")
videos
```
