# 抖音关键词与日期范围采集

只需修改下面的关键词和日期，然后运行全部单元格。遇到验证码或访问限制时程序会停止。

```python
KEYWORD = "人工智能"
START_DATE = "2026-07-01"
END_DATE = "2026-07-10"

MAX_RESULTS = 20
MAX_SCROLLS = 8
DELAY_SECONDS = 3
OUTPUT_FILE = "douyin_results.xlsx"
HEADLESS = False  # 首次运行必须保持 False，以便正常登录
INITIAL_WAIT_SECONDS = 30  # 首次打开页面后预留的登录时间
```

```python
import subprocess
import sys

command = [
    sys.executable, "-u", "main.py",
    "--keyword", KEYWORD,
    "--start-date", START_DATE,
    "--end-date", END_DATE,
    "--max-results", str(MAX_RESULTS),
    "--max-scrolls", str(MAX_SCROLLS),
    "--delay", str(DELAY_SECONDS),
    "--output", OUTPUT_FILE,
    "--initial-wait", str(INITIAL_WAIT_SECONDS),
]
if HEADLESS:
    command.append("--headless")

result = subprocess.run(
    command,
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
    timeout=1800,
)
print(result.stdout)
print(result.stderr or "无错误")
print("退出代码：", result.returncode)
if result.returncode != 0:
    raise RuntimeError("关键词爬虫运行失败，请查看上方日志")
```

```python
from pathlib import Path
import pandas as pd

output_path = Path(OUTPUT_FILE)
if output_path.exists():
    display(pd.read_excel(output_path, sheet_name="日期范围内结果"))
else:
    print("尚未生成结果文件")
```
